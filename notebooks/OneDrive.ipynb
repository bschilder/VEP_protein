`rclone` provides a unified way of syncing your files across various different cloud platforms (OneDrive, Dropbox, Google Drive, etc).

Its simplified API makes it far easier to use than the individual APIs for each of these platforms.

## 1. Install `rclone`

Some HPCs have `rclone` already installed as a module (e.g. `module load rclone`). However, this often doesn't work because a non-included dependency (`fuse`) does not get installed along with it.

In these situations, the best way to install `rclone` is via conda. 

If you don't already have conda installed, you can follow these instructions to install miniconda distribution (takes <1min).


Then, install [`rclone`](https://anaconda.org/conda-forge/rclone) and [`libfuse`](https://anaconda.org/conda-forge/libfuse) into your base conda env (or whichever env you plan on using to sync files with).


```python
%% bash

conda install -c conda-forge rclone libfuse
```

## 2. Setup OneDrive

Next we set up access to OneDrive with `rclone`. 

Follow the instructions in the subsequent prompts, and select 'Microsoft OneDrive' (by entering `28`).

See [here for full instructions](https://rclone.org/onedrive/).


- `No remotes found, make a new one?`: `n`
- `name>`: `onedrive`
- `Storage>`: `28`
- `client_id>`: `{hit ENTER}`
- `client_secret>`: `{hit ENTER}`
- `Edit advanced config?`: `n`
- `Use auto config?`: `n` (if working on a remote server)
- `config_token>`: `{"access_token":...} 
- `Option config_type.`: `1`


```python
%bash

rclone config
```

Now we need to set up a local folder where we can access the *onedrive* cloud folder.


```python
%% bash

mkdir ~/onedrive
rclone --vfs-cache-mode writes mount onedrive: ~/onedrive &
```

## 3. Upload files

Optional: For syncs that may take a long time, you may want to start a `tmux` or `screen` session first. 

This will ensure that the transfer continues even you lose connection to the server where the command is running.

```python
%% bash

tmux
```

## Download cached files

### Haplotypes

If you want to download precomputed results from OneDrive (e.g. cached haplotypes, which is much faster to download than making lots of API calls to Ensembl) you can run the same command but with the arguments reversed (OnDrive dir then your local dir).

```python
# Create the folder if it doesn't already exist
mkdir -p $HOME/.cache/ensembl_rest/haplotypes

# Download the cached files
rclone sync onedrive:VEP_protein/1KG/haplotypes $HOME/.cache/ensembl_rest/haplotypes
```
