#### Execute snakemake

Now we can run the ProHap snakemake pipeline, which will:
- automatically take in the parameters specified in the *config.yaml* file
- automatically submit multiple jobs to run different steps in parallel (if running on an HPC with a job scheduler)
- store the results in *results/hgdp* (as specified in *config.yaml*)

Read the[ ProHap docs](https://github.com/ProGenNo/ProHap/wiki/Input-&-Usage) for more information.

```python
%%bash

# Activate the snakemake conda environment and run the pipeline
conda activate snakemake ;
snakemake --jobs 30 -p --use-conda ;
```

#### Download non-standard VCF files
Annoyingly, the chrX VCF in this dataset is different by an extra ".v2." towards the end of the file name.

ProHap won't recognise the file so we need to download (and decompress) it beforehand manually.

Reported in GitHub Issue [here](https://github.com/ProGenNo/ProHap/issues/3).

```python
!wget https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/data_collections/1000G_2504_high_coverage/working/20220422_3202_phased_SNV_INDEL_SV/1kGP_high_coverage_Illumina.chrX.filtered.SNV_INDEL_SV_phased_panel.v2.vcf.gz -O $HOME/projects/ProHap/data/vcf/phased/1kGP_high_coverage_Illumina.chrX.filtered.SNV_INDEL_SV_phased_panel.vcf.gz && gunzip $HOME/projects/ProHap/data/vcf/phased/1kGP_high_coverage_Illumina.chrX.filtered.SNV_INDEL_SV_phased_panel.vcf.gz
```

#### Copy config file

```python
%%bash

scp -r $HOME/projects/VEP_protein/config/ProHap/1000_Genomes_30x_on_GRCh38.yaml $HOME/projects/ProHap/config.yaml
```

## Download precomputed haplotypes

Alternatively, you can download ProHap results precomputed by the ProHap team and shared via Zenodo.  
We provide a helper function `ph.get_zenodo` to automatic the download and caching of these files. 
Precomputed ProHap results come from the following datasets:
- 1000 Genomes Porject (1KG) results from:
         https://zenodo.org/records/12671237
- Human Reference Consortium (HRC) results from:
        https://zenodo.org/records/12671302
- Human Pangenome Reference Consortium (HPRC) results from:
        https://zenodo.org/records/12686819

**NOTE**: All of these precomputed ProHap results were generated with the parameter `phased_min_af: 0.01` which ignores variants than have an allele frequency of <1%. To get the most accurate representation of haplotypes, this should be set to `phased_min_af: 0` (as we have done in the *hgdp_config.yaml*).
